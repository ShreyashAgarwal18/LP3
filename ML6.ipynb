```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return (x + 3) ** 2

def df(x):
    return 2 * (x + 3)

print(f(2), df(2))
```

```python
x = 2
learning_rate = 0.1
history = [x]

for i in range(50):
    x = x - learning_rate * df(x)
    history.append(x)

print("Local minimum at x =", x)
print("Minimum value y =", f(x))
```

```python
x_values = np.linspace(-8, 4, 100)
history = np.array(history)

plt.plot(x_values, f(x_values), label="y = (x+3)²")
plt.scatter(history, f(history), color="red", label="Steps")
plt.title("Gradient Descent")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
